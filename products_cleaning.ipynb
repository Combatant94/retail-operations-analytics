```python
needs_case_fix = (

    products["product_name"].str.isupper() |   products["product_name"].str.match(r"[a-z]") 
)

needs_case_fix.sum()
```

```python
products.loc[needs_case_fix, "product_name"]
```

```python
products.loc[needs_case_fix,"product_name"] = products.loc[needs_case_fix,"product_name"].str.title()
```

```python
is_all_caps = products["product_name"].str.isupper()
```

```python
is_all_caps.sum()
```

```python
is_lower = products["product_name"].str.match(r"^\s*[a-z]")
```

```python
is_lower.sum()
```

```python
products["product_name"] = products["product_name"].str.replace(r"(\d)\s*g\b",
                                                                r"\1g", case = False,regex = True)
```

```python
products
```

```python
products["product_name"] = products["product_name"].str.replace(r"(\d)\s*ml\b",
                                                                r"\1ml", case = False,regex = True)
```

```python
products["product_name"] = products["product_name"].str.replace(r"(\d)(x)\s*ml\b",
                                                                r"\1ml", case = False,regex = True)
```

```python
products["product_name"] = products["product_name"].str.replace(
    r"(\d)\s*x\s*(\d)",
    r"\1 x \2",
    case=False, regex=True)
```
