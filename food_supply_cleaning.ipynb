**Inventory Consistency Validation**

***Closing Stock = Opening Stock + Deliveries − Units Sold − Expired Waste − Damaged Waste***

```python
food_supply["inventory_difference"] = (
    food_supply["opening_stock"]
    + food_supply["delivered_units"]
    - food_supply["units_sold"]
    - food_supply["waste_expired"]
    - food_supply["waste_damaged"]
    - food_supply["closing_stock"]
)

food_supply["inventory_difference"].value_counts(dropna=False).head(20)
```

```python
food_supply.loc[
    food_supply["inventory_difference"].notna()
    & (food_supply["inventory_difference"] != 0),
    [
        "date",
        "store_id",
        "sku",
        "opening_stock",
        "delivered_units",
        "units_sold",
        "waste_expired",
        "waste_damaged",
        "closing_stock",
        "inventory_difference"
    ]
].head(20)
```

```python
food_supply.loc[
    food_supply["inventory_difference"].notna()
    & (food_supply["inventory_difference"] != 0)
].groupby("store_id").size()
```

```python
food_supply["inventory_issue"] = (
    food_supply["inventory_difference"].notna()
    & (food_supply["inventory_difference"] != 0)
)

food_supply["inventory_issue"].value_counts()
```

```python
# Remove temporary validation columns
food_supply = food_supply.drop(
    columns=["inventory_difference", "inventory_issue"],
    errors="ignore"
)

# Save cleaned dataset
food_supply.to_csv("food_supply_clean.csv", index=False)

print("Cleaned dataset saved!")
print("Shape:", food_supply.shape)
```
